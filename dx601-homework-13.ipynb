```python
# DO NOT CHANGE

x3 = np.array([[-1.2556259 , -1.87605235,  0.17855308,  0.48512892, -0.63628594],
       [ 0.44304735, -0.43824678, -1.30478833, -0.11245948, -1.78622212],
       [-1.28696722, -0.49346242,  0.64796448,  0.38915627, -0.90750819],
       [-1.08362493, -0.2170256 , -1.01379518,  0.14843276, -0.25431788],
       [-0.34307836, -0.17362195,  0.77077016,  0.08711362, -0.04848665],
       [ 0.88737989,  1.30166944, -0.42145004, -0.81685573, -0.18887048],
       [ 1.97763708, -1.11463984, -0.45597113, -0.74444516,  0.72257689],
       [ 0.19500154, -1.13149454,  1.72108253,  0.28284503, -1.4325231 ],
       [-0.08665887,  0.52006977,  0.41538256, -1.53663749, -1.40234154],
       [-0.23969992, -1.66257447, -0.7795997 , -1.89104316,  0.48434458]])
```

### Problem 6

Given sample point $$y = \begin{bmatrix} 3 \\ 5 \\ 2 \end{bmatrix}$$ and principal components $$\mathbf{u}_1 = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 \\ 1 \\ 0 \end{bmatrix}, \mathbf{u}_2 = \frac{1}{\sqrt{6}} \begin{bmatrix} 1 \\ -1 \\ 2 \end{bmatrix}, \mathbf{u}_3 = \frac{1}{\sqrt{3}} \begin{bmatrix} 1 \\ -1 \\ -1 \end{bmatrix},$$ compute the coordinates of $y$.

Set `p6` to the coordinates of $y$.

```python
# DO NOT CHANGE

y6 = np.array([3, 5, 2])
u6_1 = (1 / np.sqrt(2)) * np.array([1, 1, 0])
u6_2 = (1 / np.sqrt(6)) * np.array([1, -1, 2])
u6_3 = (1 / np.sqrt(3)) * np.array([1, -1, -1])
```

```python
# YOUR CHANGES HERE

p6 = np.array([y6 @ u6_1, y6 @ u6_2, y6 @ u6_3])
```

```python
p6
```

### Problem 7

A three-dimensional data set has mean $$\mu = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}.$$
After reducing an observation to two dimensions, its PCA coordinates are $$\mathbf{z} = \begin{bmatrix} 4 \\ 2 \end{bmatrix}.$$
The retained principal components are $$\mathbf{u}_1 = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}, \mathbf{u}_2 = \frac{1}{\sqrt{2}} \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}.$$
Reconstruct the approximate observation in the original three-dimensional space.

Set `p7` to the reconstructed approximation.

```python
# DO NOT CHANGE

mu7 = np.array([1, 2, 3])
z7 = np.array([4, 2])
u7_1 = np.array([1, 0, 0])
u7_2 = (1 / np.sqrt(2)) * np.array([0, 1, 1])
```

```python
# YOUR CHANGES HERE

p7 = mu7 + z7[0] * u7_1 + z7[1] * u7_2
```

```python
p7
```

### Problem 9

A random projection from $\mathbb{R}^4$ to $\mathbb{R}^2$ is defined by $$R = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 & 1 & 0 & 0 \\ 0 & 0 & 1 & -1 \end{bmatrix}.$$
For the points $$\mathbf{x} = \begin{bmatrix} 3 \\ 1 \\ 2 \\ 0 \end{bmatrix}, \mathbf{y} = \begin{bmatrix} 1 \\ 1 \\ 0 \\ 2 \end{bmatrix},$$ set `p9` to the squared distance between their projected representations.


```python
# DO NOT CHANGE

R9 = (1 / np.sqrt(2)) * np.array([[1, 1, 0, 0], [0, 0, 1, -1]])
x9 = np.array([3, 1, 2, 0])
y9 = np.array([1, 1, 0, 2])
```

```python
# YOUR CHANGES HERE

p9 = np.sum((R9 @ x9 - R9 @ y9) ** 2)
```

```python
p9
```

### Problem 10

A random projection from $\mathbb{R}^3$ to $\mathbb{R}^2$ is defined by $$R = \frac{1}{\sqrt{2}} \begin{bmatrix} 1 & 0 & 1 \\ 0 & 1 & -1 \end{bmatrix}.$$
For $$\mathbf{x} = \begin{bmatrix} 2 \\ 1 \\ 1 \end{bmatrix},$$ set `p10` to the ratio of the projected squared length to the original squared length.

```python
# DO NOT CHANGE

R10 = (1/np.sqrt(2)) * np.array([[1, 0, 1], [0, 1, -1]])
x10 = np.array([2, 1, 1])
```

```python
# YOUR CHANGES HERE

p10 = np.sum((R10 @ x10) ** 2) / np.sum(x10 ** 2)
```

```python
p10
```

### Problem 11

Given two classes with means $$\mu_1 = \begin{bmatrix} 3 \\ 1 \end{bmatrix}, \mu_2 = \begin{bmatrix} 1 \\ 2 \end{bmatrix},$$ and within-class covariance $$\mathbf{\Sigma} = \mathbf{I},$$ compute the LDA direction $$\mathbf{w}$$ and compute the threshold $c$ for deciding between the classes, so the decision rule is $\mathbf{w}^\top > c$.

Set `p11a` to $\mathbf{w}$, and `p11b` to $c$.
Do not use normalization for either calculation; $\mathbf{w}$ should not be a unit vector.

```python
# DO NOT CHANGE

mu_11_1 = np.array([3, 1])
mu_11_2 = np.array([1, 2])
sigma_11 = np.eye(2)
```

```python
# YOUR ANSWER HERE

p11a = np.linalg.inv(sigma_11) @ (mu_11_1 - mu_11_2)
p11b = p11a @ (mu_11_1 + mu_11_2) / 2
```

```python
p11a
```

```python
p11b
```

### Problem 12

Given two classes with means $$\mu_1 = \begin{bmatrix} 3 \\ 1 \end{bmatrix}, \mu_2 = \begin{bmatrix} 1 \\ 2 \end{bmatrix},$$ and within-class covariance $$\mathbf{\Sigma} = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix},$$ compute the LDA direction $$\mathbf{w}$$ and compute the threshold $c$ for deciding between the classes, so the decision rule is $\mathbf{w}^\top > c$.

Set `p12a` to $\mathbf{w}$, and `p12b` to $c$.

```python
# YOUR ANSWER HERE

p12a = np.linalg.inv(sigma_12) @ (mu_12_1 - mu_12_2)
p12b = p12a @ (mu_12_1 + mu_12_2) / 2
```

### Problem 13

The reduced rank matrices for a truncated SVD are $$\mathbf{U}_2 = \begin{bmatrix} \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{6}} \\ \frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{6}} \\ 0 & \frac{2}{\sqrt{6}} \end{bmatrix}, \mathbf{\Sigma}_2 = \begin{bmatrix} 6 & 0 \\ 0 & 3 \end{bmatrix},$$ and $$\mathbf{V}_2^\top = \begin{bmatrix} \frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} & 0 \\ \frac{1}{\sqrt{6}} & -\frac{1}{\sqrt{6}} & -\frac{2}{\sqrt{6}} \end{bmatrix}.$$

Compute the rank-2 approximation $\mathbf{A}_2$.
Set `p13` to $\mathbf{A}_2$.

```python
# DO NOT CHANGE

U13 = np.array([[1/np.sqrt(2), 1/np.sqrt(6)], [1/np.sqrt(2), -1/np.sqrt(6)], [0, 2/np.sqrt(6)]])
S13 = np.array([[6, 0], [0, 3]])
V13 = np.array([[1/np.sqrt(2), 1/np.sqrt(6)], [1/np.sqrt(2), -1/np.sqrt(6)], [0, -2/np.sqrt(6)]])
```

```python
# YOUR CHANGES HERE

p13 = U13 @ S13 @ V13.T
```

```python
p13
```

### Problem 14

A $120 \times 80$ matrix is approximated by a rank-$k$ factorization $$\mathbf{A} \approx \mathbf{LR},$$ where $\mathbf{L}$ is $120 \times k$ and $\mathbf{R}$ is $k \times 80$.

What is the largest value of $k$ for which the factorization stores no more than 25% as many numbers as the original matrix?
Set `p14` to that $k$.

```python
# YOUR CHANGES HERE

p14 = (120 * 80) // (4 * (120 + 80))
```

```python
p14
```

### Problem 16

Let $$\mathbf{A} = \begin{bmatrix} 2 & 0 \\ 0 & 1 \end{bmatrix}.$$

Consider the rank-1 matrix $$\mathbf{B} = \begin{bmatrix} 2 & 0 \\ 1 & 0 \end{bmatrix}.$$

Set `p16` to the Frobenius norm of the approximation error between $\mathbf{A}$ and $\mathbf{B}$.
(The Frobenius norm is the square root of the sum of the squared matrix elements.)


```python
# DO NOT CHANGE

A16 = np.array([[2, 0], [0, 1]])
B16 = np.array([[2, 0], [1, 0]])
```

```python
# YOUR CHANGES HERE

p16 = np.sqrt(np.sum((A16 - B16) ** 2))
```

### Problem 17

A matrix has singular value decomposition $$\mathbf{A} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^\top,$$ where $$\mathbf{U}=\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}, \mathbf{\Sigma} = \begin{bmatrix} 4 & 0 \\ 0 & 2 \end{bmatrix}, \mathbf{V}^\top = \begin{bmatrix} \frac{3}{5} & \frac{4}{5} \\ -\frac{4}{5} & \frac{3}{5} \end{bmatrix}.$$

Set `p17` to the best rank-1 approximation of $\mathbf{A}$.

```python
# DO NOT CHANGE

U17 = np.array([[1, 0], [0, 1]])
S17 = np.array([[4, 0], [0, 2]])
V17 = np.array([[3/5, -4/5], [4/5, 3/5]])
```

```python
# YOUR CHANGES HERE

p17 = S17[0, 0] * (U17[:, :1] @ V17[:, :1].T)
```

```python
p17
```
